```python
from pydna.gel import Gel
```

```python
from pydna.genbank import genbank
```

```python
pBR322 = genbank("J01749")
```

```python
from Bio.Restriction import BglI
```

```python
pBR322_bgl = pBR322.cut(BglI)
```

```python
from pydna.gel import weight_standards, weight_standard_sample
```

```python
standard = weight_standard_sample(key="1kb+_GeneRuler")
```

```python
Gel([standard,pBR322_bgl]).run()
```
